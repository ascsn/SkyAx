```python
import pandas as pd
```

```python
def preprocess_file(input_file, output_file):
    with open(input_file, "r") as infile, open(output_file, "w") as outfile:
        for line in infile:
            parts = line.split()
            if len(parts) >= 20:  # Ensure enough columns in each row
                # Merge the first three columns into one
                first_column = " ".join(parts[:3])
                rest = parts[3:]
                new_line = f"{first_column} {' '.join(rest)}\n"
                outfile.write(new_line)

# Preprocess the file
preprocess_file("ensurf.job", "preprocessed_file.txt")

columns = ["bet2", "bet2p", "bet2n", "bet4", "bet4p", "bet4n", "energy", "epairp", "epairn", "ferp", "fern",  "iter", "rmsc", "rms", "rmsp", "rmsn", "N_boundp", "N_boundn", "last_value"]

# Reload the preprocessed file
df = pd.read_csv(
    "preprocessed_file.txt",
    delim_whitespace=True,
    header=None,
)

columns = [
    "bet2", "bet2p", "bet2n", "bet4", "bet4p", "bet4n", 
    "energy", "epairp", "epairn", "ferp", "fern", 
    "iter", "rmsc", "rms", "rmsp", "rmsn", 
    "N_boundp", "N_boundn", "last1", "last2", "last3"
]
df.columns = columns


print(df.head())
```
